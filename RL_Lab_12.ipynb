```python

import numpy as np
import pandas as pd

# 1. Initialize Parameters
gamma = 0.9
theta = 0.001
normal_reward = -1
goal_reward = 100

states = ['S0', 'S1', 'G']
actions = ['Up', 'Down', 'Left', 'Right']

V = {'S0': 0, 'S1': 0, 'G': 0}

# 2. Define Transition Rules
transitions = {
    'S0': {
        'Up': 'S0',
        'Down': 'S1',
        'Left': 'S0',
        'Right': 'S0'
    },
    'S1': {
        'Up': 'S0',
        'Down': 'S1',
        'Left': 'S1',
        'Right': 'G'
    }
}

print("========== MDP INFORMATION ==========")
print("States:", states)
print("Actions:", actions)
print("Discount Factor:", gamma)
print("Initial Values:", V)

# 3. Display Transition and Reward Information
print("\n========== TRANSITION INFORMATION ==========")

for state in ['S0', 'S1']:
    for action in actions:
        next_state = transitions[state][action]
        reward = goal_reward if next_state == 'G' else normal_reward
        print(state, action, "->", next_state, "Reward:", reward)

# 4. Value Iteration Algorithm
iteration = 0
value_history = []

while True:
    iteration += 1
    new_V = V.copy()

    for state in ['S0', 'S1']:
        q_values = []

        for action in actions:
            next_state = transitions[state][action]
            reward = goal_reward if next_state == 'G' else normal_reward

            q = reward + gamma * V[next_state]
            q_values.append(q)

        new_V[state] = max(q_values)

    delta = max(abs(new_V[s] - V[s]) for s in ['S0', 'S1'])

    value_history.append({
        'Iteration': iteration,
        'V(S0)': new_V['S0'],
        'V(S1)': new_V['S1'],
        'V(G)': 0,
        'Delta': delta
    })

    V = new_V

    if delta < theta:
        break

print("\n========== VALUE ITERATION COMPLETED ==========")
print("Total Iterations:", iteration)

# 5. Iteration-wise State Value Table
print("\n========== ITERATION-WISE STATE VALUES ==========")
df = pd.DataFrame(value_history)
display(df)

# 6. Calculate Final Q Values
Q = {}

for state in ['S0', 'S1']:
    Q[state] = {}

    for action in actions:
        next_state = transitions[state][action]
        reward = goal_reward if next_state == 'G' else normal_reward

        Q[state][action] = reward + gamma * V[next_state]

print("\n========== FINAL Q VALUE TABLE ==========")
q_table = pd.DataFrame(Q).T
display(q_table)

# 7. Display Optimal State Values
print("\n========== OPTIMAL STATE VALUES ==========")

for state in states:
    print(state, "=", V[state])

# 8. Find Optimal Policy
print("\n========== OPTIMAL POLICY ==========")

policy = {}

for state in ['S0', 'S1']:
    best_action = max(Q[state], key=Q[state].get)
    policy[state] = best_action
    print(state, "->", best_action)

# 9. Display Optimal Path
print("\n========== OPTIMAL PATH ==========")
print("S0 -> S1 -> G")

# 10. Final Summary
print("\n========== FINAL RESULTS ==========")
print("V*(S0) =", V['S0'])
print("V*(S1) =", V['S1'])
print("V*(G)  =", V['G'])
print("Optimal Action at S0:", policy['S0'])
print("Optimal Action at S1:", policy['S1'])
print("Optimal Path: S0 -> S1 -> G")
```
