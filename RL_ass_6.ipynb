```python
# REINFORCEMENT LEARNING - ASSIGNMENT 6
# Building a Random Agent for Reinforcement Learning Environments
# Environments: CliffWalking and Taxi

!pip install -q gymnasium

import gymnasium as gym
import numpy as np

NUM_EPISODES = 5


def random_agent(env_name):

    print("\n" + "=" * 60)
    print("Environment:", env_name)
    print("=" * 60)

    # Create environment
    env = gym.make(env_name)

    rewards = []

    for episode in range(1, NUM_EPISODES + 1):

        # Reset environment
        state, info = env.reset()

        total_reward = 0
        steps = 0

        while True:

            # Select random action
            action = env.action_space.sample()

            # Execute action
            next_state, reward, terminated, truncated, info = env.step(action)

            # Observe next state and reward
            state = next_state
            total_reward += reward
            steps += 1

            # Check whether episode has terminated
            if terminated or truncated:
                break

        rewards.append(total_reward)

        # Display result for each episode
        print(f"Episode {episode}: Total Reward = {total_reward}")
        print(f"Steps = {steps}")

    # Calculate average reward
    average_reward = np.mean(rewards)

    print("\nTotal Rewards:", rewards)
    print("Average Reward:", round(average_reward, 2))

    env.close()


# --------------------------------------------------
# 1. CLIFF WALKING
# --------------------------------------------------
random_agent("CliffWalking-v1")


# --------------------------------------------------
# 2. TAXI
# --------------------------------------------------
random_agent("Taxi-v4")
```
